Pip install `ultralytics`

```python
from ultralytics import YOLO
```

**Demo Detect on test images **

```python
model = YOLO('yolov8s.pt').to('cpu')

print("Run detect on two images")
results = model(source=['datasets\input\DSC_0260.jpg', 'datasets\input\DSC04251.JPG', 'Bus.jpg'], show=True, save=True)
```

**Demo video Tracking**

```python
#Tracking
model = YOLO('yolov8s.pt').to('cpu')
results = model.track(source="https://www.youtube.com/watch?v=MNn9qKG2UFI", show=True,  save=False)
```

**Demo Classify**

```python
print("Load classify model")
model = YOLO('yolov8n-cls.pt')  # load an official model
# Predict with the model
results = model(source='https://ultralytics.com/images/bus.jpg', show=True, save=True)

```
